```python
# Exam Result Checker

score = int(input("Enter your score: "))

if 70 <= score <= 100:
    print("Distinction")
elif 50 <= score <= 69:
    print("Pass")
elif 0 <= score <= 49:
    print("Fail")
else:
    print("Invalid score")
```
